```python
import pandas as pd
import networkx as nx
from networkx.algorithms import bipartite
import matplotlib.pyplot as plt
```

Graph Import and Projection

```python
edge_list = pd.read_csv('edges_5.2_(without_i_ni_si).csv')
B = nx.from_pandas_edgelist(edge_list, source='Source', target='Target')

dishes = set(edge_list['Source'])
ingredients = set(edge_list['Target'])

D = bipartite.weighted_projected_graph(B, dishes)
I = bipartite.weighted_projected_graph(B, ingredients)
```

Properties:

```python
d_edges = D.edges()
i_edges = I.edges()
```

Analysis:

```python
print(len(d_edges))
print(len(i_edges))
```
